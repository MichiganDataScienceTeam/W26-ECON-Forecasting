```python
import pandas as pd
import numpy as np

# TODO: replace with your dataset path if needed
df = pd.read_csv("your_dataset.csv")

print(df.shape)
df.head()
```

## 3️⃣ Choose target and features

Decide:
- what are we predicting? (y)
- what information do we allow the model to use? (X)

Start simple.


```python
# TODO: set target column
y = df["target"]

# TODO: choose feature columns
X = df[["feature1", "feature2", "feature3"]]
```

## 4️⃣ Time-based train/test split (CRITICAL)

For time series:

✔ train on the past  
✔ test on the future  
✘ never shuffle  

Why? To prevent **data leakage**.


```python
split = int(len(df) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]
```

## 5️⃣ Baseline model

Before using regression, create a simple baseline.

Example: predict previous value (naive forecast).

Why?
A model is only useful if it beats something simple.


```python
baseline_pred = y_test.shift(1)
baseline_pred.iloc[0] = y_train.iloc[-1]

baseline_mae = np.mean(np.abs(y_test - baseline_pred))
baseline_mae
```

## 6️⃣ Train regression model


```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)

pred = model.predict(X_test)

mae = np.mean(np.abs(y_test - pred))
mae
```

### Reflection

Did regression beat the baseline?

If not — why might that happen?


## 7️⃣ Residual diagnostics

Residual = actual − predicted

We want:
- mean near zero
- no visible pattern


```python
import matplotlib.pyplot as plt

residuals = y_test - pred

plt.scatter(pred, residuals)
plt.axhline(0)
plt.xlabel("Predicted")
plt.ylabel("Residual")
plt.show()

residuals.mean()
```

### Reflection

Do residuals show structure?

If yes, what might the model be missing?


## 8️⃣ Classification extension

Sometimes we predict categories instead of numbers.

Example:
- high vs low growth
- up vs down
- risk vs no risk

We convert regression target into classes.


```python
threshold = y_train.median()

y_train_class = (y_train > threshold).astype(int)
y_test_class = (y_test > threshold).astype(int)
```

## 9️⃣ Logistic regression

Predicts probability of class membership.

Then converts probability → class using threshold.


```python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression()
clf.fit(X_train, y_train_class)

pred_class = clf.predict(X_test)
```

## 🔟 Confusion matrix

Compares predicted vs actual classes.


```python
from sklearn.metrics import confusion_matrix, classification_report

cm = confusion_matrix(y_test_class, pred_class)
cm
```

```python
print(classification_report(y_test_class, pred_class))
```

### Reflection

Which is larger:
- false positives
- false negatives

Which matters more in a real decision context?
